# Drop or Truncate Tables

```python
SCHEMA_NAME = "inventory"
TABLES = ["InventoryTransactions", "Inventory", "Warehouses"]
```

## Truncate Tables

```python
for table in TABLES:
    spark.sql(f"TRUNCATE TABLE {SCHEMA_NAME}.{table}")
    print(f"Truncated {table}")
```

## Drop Tables

```python
for table in TABLES:
    spark.sql(f"DROP TABLE {SCHEMA_NAME}.{table}")
    print(f"Dropped {table}")
```
