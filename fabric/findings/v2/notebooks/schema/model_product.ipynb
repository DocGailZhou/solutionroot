# Product Domain Schema - v2 (Fixed)

This notebook creates the **product** schema with the following tables:

| # | Table | Description |
|---|-------|-------------|
| 1 | `product.ProductLine` | Product line dimension |
| 2 | `product.ProductCategory` | Product category hierarchy |
| 3 | `product.Product` | Core product master data |

### Fixes applied in v2
- Renamed `CategoryID` → `ProductCategoryID` for clarity and consistency
- Added `ProductLine` dimension table (new)
- Removed `BrandName` and `BrandLogoUrl` from `ProductCategory` (moved brand to Product level)
- Removed `IsoCurrencyCode` from `Product` (currency belongs in pricing/sales context)
- Standardized all date columns to ISO `DATE` format
- Added Primary Key (PK) and Foreign Key (FK) constraints on all tables

```python
spark.sql("CREATE SCHEMA IF NOT EXISTS product")
```

```python
spark.sql("""
CREATE TABLE IF NOT EXISTS product.ProductLine (
    ProductLineID INT NOT NULL,
    ProductLineName STRING NOT NULL,
    Description STRING,
    CreatedBy STRING,
    CreatedDate DATE
) USING DELTA
""")

# spark.sql("ALTER TABLE product.ProductLine ADD CONSTRAINT PK_ProductLine PRIMARY KEY (ProductLineID)")

print("✅ product.ProductLine created ")
```

```python
spark.sql("""
CREATE TABLE IF NOT EXISTS product.ProductCategory (
    ProductCategoryID STRING NOT NULL,
    ProductLineID INT,
    CategoryName STRING NOT NULL,
    CategoryDescription STRING,
    IsActive BOOLEAN,
    CreatedBy STRING,
    CreatedDate DATE
) USING DELTA
""")

# spark.sql("ALTER TABLE product.ProductCategory ADD CONSTRAINT PK_ProductCategory PRIMARY KEY (ProductCategoryID)")
# spark.sql("ALTER TABLE product.ProductCategory ADD CONSTRAINT FK_ProductCategory_ProductLine FOREIGN KEY (ProductLineID) REFERENCES product.ProductLine(ProductLineID)")

print("✅ product.ProductCategory created ")
```

```python
spark.sql("""
CREATE TABLE IF NOT EXISTS product.Product (
    ProductID STRING NOT NULL,
    ProductName STRING NOT NULL,
    ProductDescription STRING,
    BrandName STRING,
    ProductNumber STRING,
    Color STRING,
    ProductModel STRING,
    ProductCategoryID STRING,
    ProductLineID INT,
    ListPrice DECIMAL(18,2),
    StandardCost DECIMAL(18,2),
    Weight DECIMAL(18,3),
    WeightUom STRING,
    ProductStatus STRING,
    CreatedDate DATE,
    SellStartDate DATE,
    SellEndDate DATE,
    UpdatedDate DATE,
    CreatedBy STRING,
    UpdatedBy STRING
) USING DELTA
""")

# spark.sql("ALTER TABLE product.Product ADD CONSTRAINT PK_Product PRIMARY KEY (ProductID)")
# spark.sql("ALTER TABLE product.Product ADD CONSTRAINT FK_Product_Category FOREIGN KEY (ProductCategoryID) REFERENCES product.ProductCategory(ProductCategoryID)")
# spark.sql("ALTER TABLE product.Product ADD CONSTRAINT FK_Product_ProductLine FOREIGN KEY (ProductLineID) REFERENCES product.ProductLine(ProductLineID)")

print("✅ product.Product created ")
```
